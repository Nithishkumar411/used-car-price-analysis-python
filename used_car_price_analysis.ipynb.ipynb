```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the uploaded CSV dataset
df = pd.read_csv('/content/car details v4.csv')

# Display dataset information
print("Dataset loaded successfully!")
print("-" * 40)

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst five rows:")
display(df.head())
```

```python
# 1. Check the dataset size
print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])

# 2. Display all column names
print("\nColumn names:")
print(df.columns.tolist())

# 3. Check missing values in each column
print("\nMissing values:")
print(df.isnull().sum())

# 4. Count duplicate records
print("\nDuplicate rows:", df.duplicated().sum())

# 5. Check data types and non-missing values
print("\nDataset information:")
df.info()
```

```python
# STEP 5: Initial data exploration

# 1. Display five vehicle records
print("FIRST FIVE VEHICLES")
display(df.head())

# 2. Examine numerical columns
print("\nNUMERICAL SUMMARY")
display(df[['Price', 'Year', 'Kilometer']].describe())

# 3. Check the range of manufacturing years
print("\nManufacturing year range:")
print("Oldest year:", df['Year'].min())
print("Newest year:", df['Year'].max())

# 4. Check the range of listed prices
print("\nListed price range:")
print("Lowest price:", df['Price'].min())
print("Highest price:", df['Price'].max())

# 5. Examine fuel types
print("\nVehicles by fuel type:")
print(df['Fuel Type'].value_counts())

# 6. Examine transmission types
print("\nVehicles by transmission:")
print(df['Transmission'].value_counts())

# 7. Examine car manufacturers
print("\nTop 10 manufacturers by number of records:")
print(df['Make'].value_counts().head(10))
```

```python
# STEP 7: Investigate unusual records

# 1. Check the 10 most expensive vehicles
print("TOP 10 MOST EXPENSIVE VEHICLES")

display(
    df.nlargest(10, 'Price')[
        ['Make', 'Model', 'Price', 'Year', 'Kilometer']
    ]
)

# 2. Check the 10 vehicles with the highest mileage
print("\nTOP 10 VEHICLES BY KILOMETRES DRIVEN")

display(
    df.nlargest(10, 'Kilometer')[
        ['Make', 'Model', 'Price', 'Year', 'Kilometer']
    ]
)

# 3. Count vehicles priced above ₹1 crore
print("\nVehicles priced above 1 crore:")
print((df['Price'] > 10000000).sum())

# 4. Count vehicles with mileage above 5 lakh km
print("\nVehicles with mileage above 5 lakh km:")
print((df['Kilometer'] > 500000).sum())

# 5. Check for invalid or missing core values
print("\nMissing values in important columns:")
print(df[['Make', 'Model', 'Price', 'Year', 'Kilometer']].isnull().sum())

print("\nNon-positive prices:")
print((df['Price'] <= 0).sum())
```

```python
# STEP 8: Data Cleaning

# Keep the original dataset unchanged
df_clean = df.copy()

# Remove unnecessary spaces from text columns
for column in df_clean.select_dtypes(include='object').columns:
    df_clean[column] = df_clean[column].str.strip()

# Remove rows with missing values in essential analysis columns
df_clean = df_clean.dropna(
    subset=['Make', 'Model', 'Price', 'Year', 'Kilometer']
)

# Keep only valid positive prices
df_clean = df_clean[df_clean['Price'] > 0]

# Keep reasonable manufacturing years
df_clean = df_clean[
    (df_clean['Year'] >= 1980) &
    (df_clean['Year'] <= 2026)
]

# Keep non-negative mileage
df_clean = df_clean[df_clean['Kilometer'] >= 0]

# Create a separate dataset for mileage analysis.
# Exclude mileage above 500,000 km from this specific analysis only.
df_mileage = df_clean[df_clean['Kilometer'] <= 500000].copy()

# Display cleaning results
print("Original dataset rows:", len(df))
print("Cleaned dataset rows:", len(df_clean))
print("Rows available for mileage analysis:", len(df_mileage))
print("Remaining duplicate rows:", df_clean.duplicated().sum())
print("Missing essential values:")
print(df_clean[['Make', 'Model', 'Price', 'Year', 'Kilometer']].isnull().sum())
```

```python
# Average listed price by car manufacturer

brand_prices = (
    df_clean.groupby('Make')['Price']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=brand_prices.index,
    y=brand_prices.values
)

plt.title('Top 10 Manufacturers by Average Listed Car Price')
plt.xlabel('Manufacturer')
plt.ylabel('Average Listed Price (dataset units)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Average listed price by fuel type

fuel_prices = (
    df_clean.groupby('Fuel Type')['Price']
    .mean()
    .sort_values(ascending=False)
)

# Draw the chart
plt.figure(figsize=(10, 6))
sns.barplot(x=fuel_prices.index, y=fuel_prices.values)

plt.title('Average Used-Car Price by Fuel Type')
plt.xlabel('Fuel Type')
plt.ylabel('Average Listed Price (dataset units)')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

# Display average prices in a table
print("Average listed price by fuel type:")
display(fuel_prices.round(2).to_frame('Average Price'))
```

```python
# STEP 12: Compare transmission types

transmission_prices = (
    df_clean.groupby('Transmission')['Price']
    .mean()
    .sort_values(ascending=False)
)

# Create the bar chart
plt.figure(figsize=(8, 5))

sns.barplot(
    x=transmission_prices.index,
    y=transmission_prices.values
)

plt.title('Average Used-Car Price: Manual vs Automatic')
plt.xlabel('Transmission Type')
plt.ylabel('Average Listed Price (dataset units)')
plt.tight_layout()
plt.show()

# Display the exact averages
print("Average listed price by transmission:")
display(transmission_prices.round(2).to_frame('Average Price'))

# Count vehicles in each category
print("\nNumber of vehicles by transmission:")
print(df_clean['Transmission'].value_counts())
```

```python
# STEP 13: Analyze car age vs listed price

# Calculate age using the current year
current_year = 2026

df_clean['Car Age'] = current_year - df_clean['Year']

# Create a scatter plot
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_clean,
    x='Car Age',
    y='Price',
    alpha=0.6
)

plt.title('Car Age vs Listed Price')
plt.xlabel('Car Age (years)')
plt.ylabel('Listed Price (dataset units)')
plt.tight_layout()
plt.show()

# Calculate the correlation
age_correlation = df_clean['Car Age'].corr(df_clean['Price'])

print("Correlation between car age and price:", round(age_correlation, 3))
```

```python
# STEP 14: Kilometres Driven vs Listed Price

# Use df_mileage to reduce the influence of extreme mileage records
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_mileage,
    x='Kilometer',
    y='Price',
    alpha=0.6
)

plt.title('Kilometres Driven vs Listed Price')
plt.xlabel('Kilometres Driven')
plt.ylabel('Listed Price (dataset units)')
plt.tight_layout()
plt.show()

# Calculate the correlation
mileage_correlation = df_mileage['Kilometer'].corr(
    df_mileage['Price']
)

print(
    "Correlation between kilometres driven and price:",
    round(mileage_correlation, 3)
)
```

```python
# STEP 15: Final analysis tables

# 1. Calculate the correlation values
print("Car age vs price correlation:")
print(round(df_clean['Car Age'].corr(df_clean['Price']), 3))

print("\nMileage vs price correlation:")
print(round(df_mileage['Kilometer'].corr(df_mileage['Price']), 3))


# 2. Find the 10 highest-priced vehicles
print("\nTOP 10 HIGHEST-PRICED VEHICLES")

display(
    df_clean.nlargest(10, 'Price')[
        ['Make', 'Model', 'Price', 'Year', 'Kilometer']
    ]
)


# 3. Find the 10 lowest-priced vehicles
print("\nTOP 10 LOWEST-PRICED VEHICLES")

display(
    df_clean.nsmallest(10, 'Price')[
        ['Make', 'Model', 'Price', 'Year', 'Kilometer']
    ]
)


# 4. Count vehicles by manufacturer
print("\nNUMBER OF VEHICLES BY MANUFACTURER")

display(
    df_clean['Make'].value_counts().head(10).to_frame(
        'Number of Vehicles'
    )
)


# 5. Check remaining missing values
print("\nMISSING VALUES AFTER CLEANING")

print(df_clean.isnull().sum())
```
